# Graph Section 
1. Scatter Plott

```python
plt.figure(figsize=(8,5))

plt.scatter(df['cgpa'], df['iq'])

plt.title("CGPA vs IQ")
plt.xlabel("CGPA")
plt.ylabel("IQ")

plt.show()
```

# Histogram of cgpa Distribution


```python
plt.figure(figsize=(8,5))

plt.hist(df['cgpa'], bins=10)

plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")

plt.show()
```

# Line graph of cgpa

```python
plt.figure(figsize=(10,5))

plt.plot(df['cgpa'])

plt.title("CGPA Line Graph")
plt.xlabel("Students")
plt.ylabel("CGPA")

plt.show()
```

# Bar Chart of Placement Count

```python
placement_count = df['placement'].value_counts()

plt.figure(figsize=(6,5))

plt.bar(placement_count.index, placement_count.values)

plt.title("Placement Status Count")
plt.xlabel("Placement")
plt.ylabel("Count")

plt.show()
```

# Pie chart of placement

```python
placement_count = df['placement'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(placement_count.values,
        labels=['Not Placed', 'Placed'],
        autopct='%1.1f%%')

plt.title("Placement Distribution")

plt.show()
```
